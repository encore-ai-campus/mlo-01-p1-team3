```python
import requests
from bs4 import BeautifulSoup


FAQ_URL = "http://192.168.0.51:4000/faqs"


def inspect_faq_items():

    response = requests.get(
        FAQ_URL,
        timeout=10
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    # FAQ처럼 보이는 article / details / li / div 후보 확인
    candidates = soup.select(
        "article, details, li, .faq-item, .faq-card, .accordion-item"
    )

    print(f"후보 개수: {len(candidates)}")

    for i, item in enumerate(candidates[:10], start=1):

        text = item.get_text(
            " ",
            strip=True
        )

        if len(text) < 20:
            continue

        print("\n" + "=" * 60)
        print(f"후보 {i}")
        print("=" * 60)

        print(item.prettify()[:4000])


if __name__ == "__main__":
    inspect_faq_items()
```
